```python

import numpy as np
from astropy.io import fits
import matplotlib.pyplot as plt
```

```python
# --- 1) Paths dos FITS calibrados ---
fits_files = {
    "B": "m57_blue_calibrated.fits",
    "V": "m57_green_calibrated.fits",
    "R": "m57_red_calibrated.fits"
}

# --- 2) Zero points por banda (exemplo, substitui pelos teus) ---
ZP = {
    "B": 25.0,
    "V": 25.0,
    "R": 25.0
}

# --- 3) Carregar dados em arrays ---
flux = {}
for band, fname in fits_files.items():
    with fits.open(fname) as h:
        flux[band] = h[0].data.astype(float)

# --- 4) Converter fluxos em magnitudes por pixel ---
eps = 1e-12  # evita log(0)
mag = {}
for band in ["B", "V", "R"]:
    mag[band] = -2.5 * np.log10(np.maximum(flux[band], eps)) + ZP[band]

# --- 5) Mapas de cor B-V e V-R ---
BV_map = mag["B"] - mag["V"]
VR_map = mag["V"] - mag["R"]

# --- 6) Visualizar mapas de cor ---
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(BV_map, origin='lower', cmap='viridis')
plt.colorbar(label='B-V')
plt.title('Mapa de cor B-V')

plt.subplot(1,2,2)
plt.imshow(VR_map, origin='lower', cmap='viridis')
plt.colorbar(label='V-R')
plt.title('Mapa de cor V-R')

plt.tight_layout()
plt.show()
```
