Resize The Images

```python
import os
import numpy as np
from astropy.io import fits
import matplotlib.pyplot as plt
from astropy.visualization import ZScaleInterval, ImageNormalize
```

Red Filter

```python
# Diretório com a imagem
dir_red = "m57_red_stacked/"
dir_red_crop = "m57_red_resize/"
fits_file = os.path.join(dir_red, "m57_red_stacked.fits")  # mesma imagem que o nome da pasta

# Carregar a imagem
data = fits.getdata(fits_file)
header = fits.getheader(fits_file)

# Número de pixels a cortar de cada lado (ajusta conforme necessário)
crop_top = 150
crop_bottom = 150
crop_left = 70
crop_right = 70

# Cortar a imagem (crop simétrico)
cropped = data[crop_top:data.shape[0]-crop_bottom, crop_left:data.shape[1]-crop_right]

# Guardar a imagem cortada
out_file = os.path.join(dir_red_crop, "m57_red_stacked_cropped.fits")
fits.writeto(out_file, cropped, header, overwrite=True)

print("Shape final da imagem:", cropped.shape)

# Plot usando ZScale
norm = ImageNormalize(cropped, interval=ZScaleInterval())
plt.figure(figsize=(8,8))
plt.imshow(cropped, cmap="gray", origin="lower", norm=norm)
plt.colorbar(label="ADU")
plt.title("M57 Red Filter - Cropped (ZScale)")
plt.show()
```
